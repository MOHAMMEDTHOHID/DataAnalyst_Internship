```python
import pandas as pd
df=pd.read_csv('/content/customer_shopping_behavior.csv')
df.head()
```

Step 1: Import libraries and load CSV

The dataset needs to be loaded for analysis. Using pandas, we read the CSV into a DataFrame and preview it to understand its structure.

```python
print("missing values per column:\n")
df.isnull().sum()
```

Step 2: Check for missing values

Missing data can affect analysis and results. df.isnull().sum() identifies which columns have gaps so we can handle them appropriately.

```python
df['Review Rating']=df['Subscription Status'].fillna(df['Review Rating'].median)
print("missing values after cleaning:\n")
df.isnull().sum()
```

Step 3: Fill missing “Review Rating”

Missing review ratings can skew insights. We fill them with the median, which reduces the impact of outliers and preserves the overall data distribution

```python
print(f"duplicate values or row")
df.duplicated().sum()
df=df.drop_duplicates()
print("duplicate values removed")
```

Step 4: Remove duplicates

Duplicate rows can distort results. df.drop_duplicates() removes repeated entries to ensure each record is unique.

```python
categorical_cols=['Gender','Category','Location','Season']
for col in categorical_cols:
  df[col]=df[col].astype(str).str.strip().str.title()
print("\n Standard categorical columns:")
print(df[categorical_cols].head())
```

Step 5: Standardize categorical columns

Inconsistent text formatting can cause errors in analysis. We strip spaces and capitalize values to standardize columns like Gender, Category, Location, and Season.

```python

df['Age Group'] =pd.cut(df['Age'],bins=[0,30,50,100] , labels =['Young','Adult','Senior'])
print("\nage group distribution:")
print(df['Age Group'].value_counts())
```

Step 6: Create age groups

Continuous age values are hard to analyze in segments. Using pd.cut(), we group ages into Young, Adult, and Senior categories for easier analysis.

```python

df.to_csv('cleaned_data_csv',index=False)
print("\nProcess complete, File saved as 'cleaned_data_csv'.")
df.head()
```

Step 7: Save the cleaned data

After cleaning, saving the dataset prevents repeating the process. df.to_csv() exports the cleaned, standardized data for future use.
